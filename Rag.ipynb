```python
from langchain_community.document_loaders import PyMuPDFLoader
```

```python
#pdf loder 
loder=PyMuPDFLoader(r"C:/Users/intel/Desktop/Gen_Lab/data/PowerBI_Day1.pptx.pdf")
docs=loder.load()
print(len(docs))
print(docs[0].page_content[:300])#preview 300 char 
print(docs[0].metadata)#  example siurse page 0 
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

spiliter=RecursiveCharacterTextSplitter(
    chunk_size=500, 
    chunk_overlap=50)

chunks=spiliter.split_documents(docs)
print(len(chunks))
print(chunks[0].page_content) 
```

```python
from langchain_ollama import OllamaEmbeddings

embeddings=OllamaEmbeddings(model="nomic-embed-text")

#quick test 
vector=embeddings.embed_query("what is sql ")
print(len(vector))
print(vector[:5])#preview first 5 number 
```

```python
from langchain_ollama import ChatOllama

llm=ChatOllama(model="llama3.2:3b")
```

```python
from langchain_classic.memory import ConversationBufferMemory

memory=ConversationBufferMemory( 
    memory_key="chat_history",
    return_messages=True
)
```

```python
from langchain_classic.chains import ConversationalRetrievalChain

qa_chain=ConversationalRetrievalChain.from_llm( 
    llm,
    retriever=retriever,
    memory=memory

)
```

```python


while True:
    question=input("you")
    if question.lower()=="exit":
        break
    result=qa_chain.invoke({"question":question})
    print("AI",result["answer"])
```

```python

from langchain_community.vectorstores import FAISS

vectorstore=FAISS.from_documents(chunks,embeddings) 

vectorstore.save_local("faiss_index") 
```

```python
retriever=vectorstore.as_retriever(
    search_kwargs={"k":3}

)

results=retriever.invoke("where is AI used")
for i in results:
    print(i.page_content[:50],"....\n")
```
