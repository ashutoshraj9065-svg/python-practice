```python
def bubbl(a): 
    n=len(a)
    
    for i in range(n):
        count=0
        for j in range(0,n-i-1):
            if a[j]>a[j+1]:
                a[j],a[j+1]=a[j+1],a[j]
                count=count+1
        if count==0:
            break
    print(count)
        
    return a



        
                
            
    
```

```python
k=[1,2,3,4,5]
bubbl(k)
```
