```python
nums = [4, 7, 2, 9, 5] 

largst=nums[0] 

for i in nums: 
    if i>largst: 
        largst=i 
print(largst) 
```

```python
nums = [4, 7, 2, 9, 5] 
largest=nums[0] 
second=float('-inf')

for i in nums:
    if i>largest:
        second=largest
        largest=i 
    elif i>second:
        second=i
        
print(largest) 
print(second) 
```
