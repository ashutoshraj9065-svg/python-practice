```python
numbers = [2, 5, 8, 9, 12, 15] 

even=[a for a in numbers if a%2==0]
even
```

```python
number = [1, 2, 3, 4, 5]

odd_even=[ "even" if a%2==0 else "odd" for a in number] 
odd_even
```

```python
numbers = [1, 2, 3, 4, 5, 6] 

sq_cube=[a**2 if a%2==0 else a**3 for a in numbers]
sq_cube
```
