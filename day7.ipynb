```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python


df = pd.DataFrame({
    "name": ["A", "B", "C", "D", "E"],
    "dept": ["Sales", "Sales", "HR", "HR", "HR"],
    "salary": [90000, 70000, 80000, 60000, 50000]
})
df
df["avg_salary"] = df.groupby("dept")["salary"].transform("mean")
df
df["rank"] = df.groupby("dept")["salary"].rank(method="dense",ascending=False)
df
                                
```

```python
employee=df[df["salary"]>70000]
employee
```

```python
sales_employee = df[
    (df["dept"] == "Sales") &
    (df["salary"] > 70000)
]
sales_employee
```

```python
sale_hr_result = df[
    ((df["dept"] == "Sales") | (df["dept"] == "HR")) &
    (df["salary"] > 70000)
]
sale_hr_result

sale_hr_result2=df[(df["dept"].isin(["Sales","HR"]))& (df["salary"]>70000)]
sale_hr_result2
```
