```python
def merge(left,right):
    n=[]
    i,j=0,0
    while i<len(left) and j<len(right):
        if left[i]<right[j]:
            n.append(left[i])
            i=i+1
        else: 
            n.append(right[j])
            j=j+1
    n.extend(left[i:]) 
    n.extend(right[j:]) 
    return n 
    
```

```python
def merge_sort(arr):
    if len(arr)<=1: 
        return arr 
    mid=len(arr)//2
    left=arr[:mid]
    right=arr[mid:]
    left=merge_sort(left) 
    right=merge_sort(right)
    return merge(left,right)
```

```python
k=[6,9,2,1,4,7] 
merge_sort(k)
```
