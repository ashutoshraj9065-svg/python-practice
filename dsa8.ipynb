```python
def ashu(arr):
    n=len(arr)
    left=0
    right=3
    while left<right:
        if arr[left]%2==0:
            left=left+1
        elif arr[right]%2!=0:
            right=right-1
        else:
            arr[left],arr[right]=arr[right],arr[left]
    return arr
```
