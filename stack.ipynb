```python
# constructor it is a special type of function which is used to allocated the memory of an object 


class mystack:
    def __init__(self):
        self.st=[] 
    def pushs (self,data):
        self.st.append(data)
    def tops(self):
        if len(self.st)==0: 
            return "empty array" 
        return self.st[-1]

    def deletes(self):
        if len(self.st)==0: 
            return "under flow" 
        d=self.tops()
        self.st.pop() 
        return d 
        
```

```python
obj=mystack()
obj.pushs(20) 
obj.pushs(30) 
obj.pushs(40) 
obj.pushs(50)
obj.tops() 
obj.deletes()
obj.tops()

# for i in obj.st: 
#     print(i)
obj.st
```
